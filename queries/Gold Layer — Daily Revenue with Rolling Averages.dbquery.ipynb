```sql
-- Gold Layer: Daily Revenue with 7-day Rolling Average and Day-over-Day Delta
-- Table: lab_uc.gold.daily_revenue
-- Demonstrates: PARTITION BY, window frames (ROWS BETWEEN), LAG(), AVG() OVER()

SELECT
    event_date,
    currency,
    orders,
    revenue,
    avg_order_value,
    -- 7-day rolling average of revenue
    AVG(revenue) OVER (
        PARTITION BY currency
        ORDER BY event_date
        ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
    ) AS revenue_7day_avg,
    -- Day-over-day revenue change
    revenue - LAG(revenue) OVER (
        PARTITION BY currency
        ORDER BY event_date
    ) AS revenue_delta
FROM lab_uc.gold.daily_revenue
WHERE currency = 'USD'
ORDER BY event_date DESC
LIMIT 20;
```
