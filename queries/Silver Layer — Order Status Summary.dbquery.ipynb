```sql
-- Silver Layer: Order Status Summary
-- Table: lab_uc.silver.orders_sql
-- Demonstrates: GROUP BY, COUNT, SUM, AVG, MIN, MAX, ROUND, ORDER BY

SELECT
    status,
    COUNT(*) AS order_count,
    ROUND(SUM(amount), 2) AS total_amount,
    ROUND(AVG(amount), 2) AS avg_amount,
    ROUND(MIN(amount), 2) AS min_amount,
    ROUND(MAX(amount), 2) AS max_amount
FROM lab_uc.silver.orders_sql
GROUP BY status
ORDER BY total_amount DESC;
```
