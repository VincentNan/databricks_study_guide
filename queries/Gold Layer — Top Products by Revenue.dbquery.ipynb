```sql
-- Gold Layer: Top Products by Revenue with Ranking
-- Table: lab_uc.gold.study_product_sales
-- Demonstrates: RANK() OVER(), derived columns, ROUND

SELECT
    product_id,
    units_sold,
    revenue,
    ROUND(revenue / units_sold, 2) AS revenue_per_unit,
    RANK() OVER (ORDER BY revenue DESC) AS revenue_rank
FROM lab_uc.gold.study_product_sales
ORDER BY revenue DESC
LIMIT 10;
```
